```python
# import libraries
import json

# import os
import polars as pl

# import polars.selectors as cs
import requests

# from sqlalchemy import create_engine
```

```python
# what dish are we looking for?
dish_name = "lasagna"
cuisine_type = "italian"
```

```python
# load credentials

f = open("../secrets/edamam.json", "r")
cred = json.load(f)
f.close()

cred_appid = cred["id"]
cred_appkey = cred["key"]
```

```python
# make request to Edamam API
query_base = "https://api.edamam.com/api/recipes/v2?type=public&"

api_call = f"{query_base}q={dish_name}&app_id={cred_appid}&app_key={cred_appkey}"

resp = requests.get(api_call)
```

```python
resp
```

```python
resp.json()
```

```python
resp_dict_hits = resp.json()["hits"]
resp_dict_hits
```

```python
urls = []
labels = []
sources = []
ingreds = []
cuisines = []

for recipe in resp_dict_hits:
    recipe_path = recipe["recipe"]
    urls.append(recipe_path["url"])
    labels.append(recipe_path["label"])
    sources.append(recipe_path["source"])
    ingreds.append(recipe_path["ingredientLines"])
    cuisines.append(recipe_path["cuisineType"])

all_recipes = {
    "url": urls,
    "label": labels,
    "source": sources,
    "ingredients": ingreds,
    "cuisines": cuisines,
}

recipe_df = pl.DataFrame(all_recipes)
```

```python
recipe_df
```

```python
one_recipe = []

for listing in recipe_df["ingredients"]:
    for ingred in listing:
        one_recipe.append(ingred.lower())

one_recipe = list(set(one_recipe))

query_df = pl.DataFrame(
    data={
        "name": dish_name,
        "ingredients": [one_recipe],
        "cuisine": cuisine_type,
    }
)

query_df
```
